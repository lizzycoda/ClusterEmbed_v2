```python
root_path = '../../ne_spectrum_scRNAseq/data/'
X, labels, d = load_human(root_path=root_path)
X.shape, labels.shape
```

```python
def convert(y):
    mapping = {'EB':0, 'Neuroectoderm':1, 'Neuroepithelium':2, 'Organoid-1M':3,
       'Organoid-2M':4, 'Organoid-4M':5, 'iPSCs':6}
    return [mapping[y[i]] for i in range(len(y))]
labels = convert(labels)
labels = np.array(labels)
```

# Isomap + Geodesic distances embedding

## Cluster data with Leiden algorithm

```python
st = time.time()
adata = sc.AnnData(X)
sc.pp.neighbors(adata, n_neighbors=15)
time.time()-st
```

```python
st = time.time()
sc.tl.leiden(adata, resolution=0.15) 
clusters = adata.obs['leiden']
print(time.time()-st)
rand_score(clusters, labels)
```

```python
clusters = np.array(clusters).astype(int)
confusion_matrix(clusters, labels)[:,0:7]
```

```python
np.unique(clusters, return_counts = True)
```

## Embed each cluster

```python
st = time.time()
l_isomap  = LandmarkIsomap(n_components = 2, n_landmarks = 300)
embeddings = np.zeros((n,2))
for i in np.unique(clusters):

    cc = np.sum(clusters == i)
    if cc > 300:
        cluster_emb = l_isomap.fit_transform(adata[clusters == i])
        embeddings[clusters == i] = cluster_emb - np.mean(cluster_emb, axis = 0)
    else:
        l_isomap_temp = LandmarkIsomap(n_components = 2, n_landmarks = cc)
        cluster_emb = l_isomap_temp.fit_transform(adata[clusters == i])
        embeddings[clusters == i] = cluster_emb - np.mean(cluster_emb, axis = 0)
time.time()-st
```

## Align

```python
st = time.time()
# approximate geodesic distances 
alignment_idxs = np.random.choice(n, 5000, replace = False)

D = shortest_path(
    adata.obsp["distances"],
    directed=False,
    indices= alignment_idxs # only compute from these sources
)

embedding_sampled = embeddings[alignment_idxs]
clusters_sampled = clusters[alignment_idxs]
D = D[:,alignment_idxs]
D[np.isinf(D)] =  np.max(D[np.isfinite(D)]) # replace infinite values

#reformat
distances = {} 

n_clusters = len(np.unique(clusters))
for i in range(n_clusters):
     for j in range(i+1, n_clusters):
            distances[i,j] = D[clusters_sampled == i, :][:,clusters_sampled == j]
time.time()-st
```

```python
st = time.time()
out = Embedding(embedding_sampled, clusters_sampled, distances, alpha =1)
ce_1 = out.align_embedding(m = 50)
ce_1 =  out.embed_points(embeddings,clusters)
time.time()-st
```

```python
out.get_default_alpha()
```

```python
st = time.time()
out = Embedding(embedding_sampled, clusters_sampled, distances, alpha =1.45)
ce_11 = out.align_embedding(m = 50)
ce_11 =  out.embed_points(embeddings,clusters)
time.time()-st
```

## Also fit with trimap embeddings

```python
import time
```

```python

st = time.time()
trimap_embeddings =  np.load("../results/brain/trimap_clusters.npy", allow_pickle=False)

# rescale the embeddings:
for i in range(10):
    scaling_idxs = clusters[alignment_idxs] == i  #T/F on the alignment idxs if in cluster i
    orig_pdist = squareform(D[scaling_idxs,:][:,scaling_idxs])
    emb_pdist = pdist(trimap_embeddings[alignment_idxs[scaling_idxs]])
    scale  = np.sum(orig_pdist*emb_pdist)/np.sum(emb_pdist**2)
    trimap_embeddings[clusters == i] *= scale
time.time()-st
```
